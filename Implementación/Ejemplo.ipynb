# Documentación de uso del modelo de productividad agrícola

Esta sección describe cómo utilizar el modelo entrenado para predecir la productividad agrícola con datos reales, qué archivos se requieren y cómo debe ser la estructura de los datos de entrada.

## Archivos requeridos para la predicción
- **modelo_random_forest.joblib**: Modelo entrenado de RandomForestRegressor.
- **scaler.joblib**: Objeto StandardScaler para estandarizar los datos.
- **pca.joblib**: Objeto PCA para reducción de dimensionalidad.
- **uso_modelo_agricola.py**: Script de ejemplo para cargar el modelo y hacer predicciones.

Todos estos archivos deben estar en la misma carpeta o debes ajustar las rutas en el script.

## Parámetros de entrada requeridos
El modelo espera los siguientes parámetros (columnas) en el mismo orden y con los mismos nombres:

- Anio
- areaSembradaHa
- areaCosechadaHa
- produccionTon
- CicloPermanente
- CicloTransitorio
- GpCereales
- grupoCultivo_Cultivos para condimentos, bebidas medicinales y aromáticas
- grupoCultivo_Cultivos tropicales tradicionales
- GpFrutales
- GpHortalizas
- GpLeguminosas
- GpOleaginosas
- grupoCultivo_Raíces y tubérculos
- Boyaca
- EdoCañaVerde
- Edofresco
- EdoFibraCabuya
- EdoGrano
- EdoGranoSeco
- EdoPaddyCascaraVerde
- EdoSecoTrilla
- Periodo_A
- Periodo_B
- Periodo_C

**Nota:** Si falta alguna columna o el nombre no coincide, el modelo no funcionará correctamente.

## Ejemplo de uso práctico (script Python)
El archivo `uso_modelo_agricola.py` contiene un ejemplo completo de cómo cargar los archivos del modelo y realizar una predicción. Puedes modificar los valores de entrada según tus datos reales.

```python
import joblib
import pandas as pd

# Cargar los archivos generados
rf_model = joblib.load('modelo_random_forest.joblib')
scaler = joblib.load('scaler.joblib')
pca = joblib.load('pca.joblib')

# Definir los datos de entrada EXACTAMENTE con los nombres y orden de las columnas de X
entrada_usuario = {
    'Anio': [2025],
    'areaSembradaHa': [2.5],
    'areaCosechadaHa': [2.5],
    'produccionTon': [7.0],
    'CicloPermanente': [0],
    'CicloTransitorio': [1],
    'GpCereales': [0],
    'grupoCultivo_Cultivos para condimentos, bebidas medicinales y aromáticas': [0],
    'grupoCultivo_Cultivos tropicales tradicionales': [0],
    'GpFrutales': [1],
    'GpHortalizas': [0],
    'GpLeguminosas': [0],
    'GpOleaginosas': [0],
    'grupoCultivo_Raíces y tubérculos': [0],
    'Boyaca': [1],
    'EdoCañaVerde': [0],
    'Edofresco': [0],
    'EdoFibraCabuya': [0],
    'EdoGrano': [0],
    'EdoGranoSeco': [0],
    'EdoPaddyCascaraVerde': [0],
    'EdoSecoTrilla': [0],
    'Periodo_A': [0],
    'Periodo_B': [1],
    'Periodo_C': [0]
}

X_usuario = pd.DataFrame(entrada_usuario)

# Estandarizar y aplicar PCA
datos_scaled = scaler.transform(X_usuario)
datos_pca = pca.transform(datos_scaled)

# Realizar la predicción
prediccion = rf_model.predict(datos_pca)
print(f'Predicción de rendimiento (ton/ha): {prediccion[0]:.2f}')
```

## Flujo de uso para usuarios reales
1. El usuario recopila los datos requeridos y los ingresa en el formato adecuado.
2. El sistema carga los archivos del modelo (`.joblib`), estandariza y transforma los datos.
3. Se realiza la predicción y se entrega el resultado al usuario.
4. El usuario puede tomar decisiones informadas y reportar resultados para retroalimentar el sistema.
